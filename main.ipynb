```python
# Define the place where the Docs are and extentions
docs_directory = "Docs"
supported_extensions = [".txt", ".py"]
```

```python
# append all files names and paths together

files_path = []

for filename in os.listdir(docs_directory):
    file_path = os.path.join(docs_directory, filename)
    if os.path.isfile(file_path):
        file_extention = os.path.splitext(filename)[1].lower()
        if file_extention in supported_extensions:
            files_path.append(file_path)


print(f"the files are {files_path}")
```
